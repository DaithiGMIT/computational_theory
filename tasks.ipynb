### Understanding Bit Shifting

Bit shifting operations are fundamental in binary manipulation and crucial for this task.  

- The **left shift operator** `<<` moves all bits of a number to the left by a specified number of positions.  
  Each left shift is equivalent to multiplying the number by 2.
  
- The **right shift operator** `>>` moves all bits of a number to the right by a specified number of positions.  
  Each right shift is approximately equivalent to integer division by 2.

When performing rotations in a fixed-size bit field , it is important to also mask the result with `0xFFFFFFFF` to ensure that bits do not overflow beyond 32 bits.

For example:
- `5 << 1` shifts the bits of `5` (binary `000...0101`) left by 1 position, resulting in `10`.
- `8 >> 1` shifts the bits of `8` (binary `000...1000`) right by 1 position, resulting in `4`.


## rotl(x, n=1)

The `rotl` function rotates the bits of a 32-bit unsigned integer `x` to the left by `n` positions.  
Bits that are shifted out from the left are reinserted on the right.  
A bitmask of `0xFFFFFFFF` is applied to ensure the result remains a 32-bit unsigned integer.


````python
def rotl(x, n=1):
    """Rotate bits of a 32-bit unsigned integer to the left by n positions."""
    
    # Shift x left by n bits
    left_shifted = (x << n)
    
    # Shift x right by (32 - n) bits
    right_shifted = (x >> (32 - n))
    
    # Combine the shifted values
    combined = left_shifted | right_shifted
    
    # Mask to 32 bits
    result = combined & 0xFFFFFFFF
    
    return result
````

### Helper Functions

The following helper functions are used to improve the clarity of output and to verify results during testing.


````python
def to_32bit_bin(val):
    """Return a 32-bit binary string with 0b prefix."""
    return format(val, '#034b')
````

````python
def verify_output(actual, expected):
    """Compare actual and expected values and print a verification result."""
    if actual == expected:
        print("Output matches expected result.")
    else:
        print("Output does not match expected result.")
````

### Testing rotl(x, n=1)

Each example includes an expected output description to verify correctness.



#### Example 1

**Expected:**
- The single `1` bit moves to the second position.
- Result full binary (32 bits): `0b00000000000000000000000000000010`
- Shortened binary: `0b10`
- Result decimal: `2`


````python
# Test Example 1
x = 0b00000000000000000000000000000001
rotated = rotl(x, 1)

print(f"Input (full binary): {to_32bit_bin(x)}")
print(f"Input (shortened binary): {bin(x)}")
print(f"Output after rotl(x, 1) (full binary): {to_32bit_bin(rotated)}")
print(f"Output after rotl(x, 1) (shortened binary): {bin(rotated)}")
print(f"Output after rotl(x, 1) (decimal): {rotated}")

# Verify result
expected = 0b10
verify_output(rotated, expected)
````
