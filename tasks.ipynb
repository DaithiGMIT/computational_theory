### Understanding Bitwise Operations

Bitwise operations are fundamental to binary manipulation, especially in cryptographic algorithms like SHA-256.  
The following operators are used throughout the binary representations task:

- **Left Shift (`<<`)**  
  Moves all bits to the left by a specified number of positions.  
  Each left shift multiplies the number by `2`.  
  Example: `5 << 1` → `10`.

- **Right Shift (`>>`)**  
  Moves all bits to the right by a specified number of positions.  
  Each right shift approximately divides the number by `2`.  
  Example: `8 >> 1` → `4`.

- **Bitwise AND (`&`)**  
  Compares each bit of two numbers and returns `1` if both bits are `1`, otherwise returns `0`.  
  Example: `0b1100 & 0b1010` → `0b1000`.

- **Bitwise OR (`|`)**  
  Compares each bit of two numbers and returns `1` if at least one of the bits is `1`.  
  Example: `0b1100 | 0b1010` → `0b1110`.

- **Bitwise XOR (`^`)**  
  Compares each bit of two numbers and returns `1` if the bits are different, otherwise returns `0`.  
  Example: `0b1100 ^ 0b1010` → `0b0110`.

- **Bitwise NOT (`~`)**  
  Inverts each bit of the number: `1` becomes `0`, and `0` becomes `1`.  
  Example: `~0b00001111` → `0b11110000` (after considering 32-bit masking).

- **Masking with `0xFFFFFFFF`**  
  After shifting or other operations, numbers are masked to stay within 32 bits by using bitwise AND with `0xFFFFFFFF`.  
  Example: `(result) & 0xFFFFFFFF`.

These operations are essential for manipulating binary data accurately and efficiently.


## rotl(x, n=1)

The `rotl` function rotates the bits of a 32-bit unsigned integer `x` to the left by `n` positions.  
Bits that are shifted out from the left are reinserted on the right.  
A bitmask of `0xFFFFFFFF` is applied to ensure the result remains a 32-bit unsigned integer.


````python
def rotl(x, n=1):
    """Rotate bits of a 32-bit unsigned integer to the left by n positions."""
    
    # Shift x left by n bits
    left_shifted = (x << n)
    
    # Shift x right by (32 - n) bits
    right_shifted = (x >> (32 - n))
    
    # Combine the shifted values
    combined = left_shifted | right_shifted
    
    # Mask to 32 bits
    result = combined & 0xFFFFFFFF
    
    return result
````

## `rotr(x, n=1)`

The `rotr` function rotates the bits of a 32-bit unsigned integer `x` to the right by `n` positions.  
Bits that are shifted out from the right are reinserted at the left.  
A bitmask of `0xFFFFFFFF` is applied to ensure the result remains a 32-bit unsigned integer.


````python
def rotr(x, n=1):
    """Rotate bits of a 32-bit unsigned integer to the right by n positions."""
    
    # Shift x right by n bits
    right_shifted = (x >> n)
    
    # Shift x left by (32 - n) bits
    left_shifted = (x << (32 - n))
    
    # Combine the shifted values
    combined = right_shifted | left_shifted
    
    # Mask to 32 bits
    result = combined & 0xFFFFFFFF
    
    return result
````

## `ch(x, y, z)`

The `ch` function chooses bits from `y` or `z` based on the bits of `x`.

For each bit position:
- If the corresponding bit in `x` is `1`, the bit from `y` is chosen.
- If the corresponding bit in `x` is `0`, the bit from `z` is chosen.

This operation is fundamental in cryptographic hash functions like SHA-256.

````python
def ch(x, y, z):
    """Choose bits from y or z based on x: if bit in x is 1, take bit from y; else take bit from z."""
    
    # Select bits from y where x has 1s
    y_bits = x & y
    
    # Select bits from z where x has 0s
    z_bits = (~x) & z
    
    # Combine the results
    result = y_bits ^ z_bits
    
    return result
````

## `maj(x, y, z)`

The `maj` function computes the majority value for each bit among the three inputs `x`, `y`, and `z`.

For each bit position:
- If **at least two** of the corresponding bits in `x`, `y`, and `z` are `1`, the output bit is set to `1`.
- Otherwise, the output bit is set to `0`.

This operation is used in cryptographic algorithms like SHA-256 to combine multiple inputs securely.


````python
def maj(x, y, z):
    """Compute the majority for each bit position among x, y, and z."""
    
    # Bits where both x and y are 1
    xy = x & y
    
    # Bits where both x and z are 1
    xz = x & z
    
    # Bits where both y and z are 1
    yz = y & z
    
    # Combine all majority cases
    result = (xy | xz | yz)
    
    return result
````

  - Otherwise, the output bit is `0`.

**Expected result full binary:** `0b11110000`  
**Expected result decimal:** `240`


````python
# Test maj Example 1
x = 0b11110000
y = 0b10101010
z = 0b01010101
majority = maj(x, y, z)

print(f"x (full binary): {to_32bit_bin(x)}")
print(f"y (full binary): {to_32bit_bin(y)}")
print(f"z (full binary): {to_32bit_bin(z)}")
print(f"Output (full binary): {to_32bit_bin(majority)}")
print(f"Output (shortened binary): {bin(majority)}")
print(f"Output (decimal): {majority}")

# Verify result
expected = 0b11110000
verify_output(majority, expected)
````

## Understanding the Original C Hash Function

The provided C function computes a simple hash value for a given string.  
Here is what the function does, step-by-step:

1. **Initialize** a variable `hashval` to `0`.
2. **Loop** through each character `*s` in the string until the null terminator (`'\0'`) is reached.
3. **For each character**:
   - Multiply the current `hashval` by `31`.
   - Add the ASCII value of the current character (`*s`) to `hashval`.
4. **After the loop**:
   - Return the final `hashval` modulo `101`.

This hash function is simple but effective for small-scale applications.


## Translating the C Function to Python

The goal is to write a Python function that behaves identically to the original C version.

Key translation points:
- In C, strings are arrays of characters terminated by a `'\0'` null character.  
  In Python, strings are automatically managed and do not require a null terminator.
- In C, characters like `*s` are accessed by pointer dereferencing.  
  In Python, we simply loop through the string directly: `for c in s:`.
- In C, `*s` gives the ASCII code implicitly.  
  In Python, the `ord(c)` function is used to get the ASCII integer value of character `c`.
- The mathematical operations are the same:
  - Multiply the current hash value by 31.
  - Add the ASCII value of the current character.
- After processing all characters, return `hashval % 101` to constrain the result to a small range.



````python
def hash_string(s):
    """Compute the hash value of a string using a polynomial rolling hash function."""
    
    # Initialize the hash value to zero
    hashval = 0
    
    # Debug: Print the initial hash value
    print(f"Initial hashval: {hashval}")
    
    # Process each character in the input string
    for c in s:
        # Store the current hash value before updating (for debugging)
        old_hashval = hashval
        
        # Update hash by multiplying current hash by 31 and adding the ASCII value
        # This creates a polynomial hash where each character contributes to the result
        hashval = ord(c) + 31 * hashval
        
        # Debug: Print each step of the calculation with detailed breakdown
        print(f"Character '{c}' (ASCII {ord(c)}): {ord(c)} + 31 * {old_hashval} = {hashval}")
    
    # Calculate the final result using modulo 101 to keep the hash in a fixed range
    result = hashval % 101
    
    # Debug: Print the final hash value before and after modulo operation
    print(f"Final hashval: {hashval}")
    print(f"After modulo 101: {result}")
    
    # Return the hash value in the range [0, 100]
    return result
````

## Why 31 and 101?

In the original C hash function, the constants 31 and 101 were chosen for specific reasons:

- **31 (Multiplier):**
  - 31 is a small prime number.
  - Multiplying by a prime helps to distribute the hash values more uniformly.
  - 31 is particularly efficient in binary computation because multiplication by 31 can be optimized as `(x << 5) - x` (shift and subtract), making it fast.

- **101 (Modulo Base):**
  - 101 is a prime number.
  - Using a prime number as the modulo base reduces the chance of hash collisions (different inputs giving the same hash).
  - 101 keeps the final hash value within a small range (0–100), making it suitable for small hash tables.

Together, these choices balance **speed**, **simplicity**, and **good distribution** of hash values, especially for educational examples and small programs.

## Understanding SHA256 Padding

In SHA256, messages must be padded before hashing to ensure the total length is a multiple of 512 bits.  
The padding process follows these steps:

1. **Append a single `1` bit**:  
   - In binary, this is `10000000` (one `1` bit followed by seven `0` bits).
   - In hexadecimal, it is `0x80`.
   
2. **Append enough `0` bits**:  
   - Add zero (`0x00`) bytes until the total length is **64 bits (8 bytes) less than a multiple of 512**.
   - This makes room for appending the message length at the end.

3. **Append the original message length**:  
   - Represent the original message length **in bits**, not bytes.
   - Encode it as a **64-bit big-endian integer** (8 bytes).

The final padded message length will always be a multiple of 512 bits (or 64 bytes).
This padding ensures the message structure is predictable and secure for the SHA256 algorithm.

- Appends enough `0x00` bytes.
- Appends the original length (in bits) encoded as a big-endian 64-bit integer.


````python
def calculate_sha256_padding_from_bytes(message_bytes):
    """Given message bytes, calculate the SHA256 padding and return it."""
    
    # Calculate the original message length in bits
    original_length_bits = len(message_bytes) * 8
    
    # Start padding with a single '1' bit followed by 7 zeros (0x80 byte)
    padding = b'\x80'
    
    # Calculate how many zero bytes are needed
    # The total message length (including 0x80 byte) must be 448 mod 512
    total_length_bits = original_length_bits + 8  # 8 bits added by 0x80
    padding_zeros_bits = (448 - total_length_bits % 512) % 512
    padding_zeros_bytes = padding_zeros_bits // 8
    
    # Add the calculated number of zero bytes
    padding += b'\x00' * padding_zeros_bytes
    
    # Append the original message length as a 64-bit big-endian integer
    length_bytes = original_length_bits.to_bytes(8, 'big')
    padding += length_bytes
    
    return padding
````

### Helper Function: verify_padding_length

This function checks that the total length of the original message plus the padding  
is a multiple of 512 bits, as required by SHA256.


````python
def verify_padding_length(message_bytes, padding_bytes):
    """Verify that the total length (original + padding) is a multiple of 512 bits."""
    
    # Calculate total message + padding length in bits
    total_bits = (len(message_bytes) + len(padding_bytes)) * 8
    
    # Check if total length is a multiple of 512 bits
    if total_bits % 512 == 0:
        print("Total length after padding is a multiple of 512 bits.")
    else:
        print("Total length after padding is NOT a multiple of 512 bits.")
````

### Main Function: calculate_sha256_padding

This function reads a file in binary mode, calculates its SHA256 padding,  
prints the padding in hexadecimal format, and verifies that the final message length  
(including padding) is a multiple of 512 bits.


````python
def calculate_sha256_padding(filepath):
    """Read file, calculate SHA256 padding, print hex values, and verify length."""
    
    # Step 1: Read the file contents as bytes
    with open(filepath, 'rb') as f:
        message = f.read()
    
    # Step 2: Calculate the padding for the file contents
    padding = calculate_sha256_padding_from_bytes(message)
    
    # Step 3: Print the calculated padding in hexadecimal format
    print("Padding bytes (hex):")
    print(' '.join(f'{byte:02x}' for byte in padding))
    
    # Step 4: Verify that the total padded message length is correct
    verify_padding_length(message, padding)
````

## Prime Numbers Definition

A **prime number** is a natural number greater than 1 that has no divisors other than 1 and itself.  
This means a prime number cannot be formed by multiplying two smaller natural numbers.

- The number 2 is the smallest and only even prime number.
- Examples of prime numbers: 2, 3, 5, 7, 11, 13, 17, etc.

Non-prime numbers are called **composite numbers**, meaning they have factors other than 1 and themselves.


## Relevance and Uses of Prime Numbers

Prime numbers are fundamental in many areas of mathematics and computer science, particularly in:

- **Cryptography**  
  Prime numbers are used in algorithms like RSA encryption, where security depends on the difficulty of factoring large numbers into primes.

- **Hash Functions**  
  Some hashing techniques rely on prime numbers to distribute hash values more uniformly.

Understanding and generating prime numbers is essential for both theoretical and practical applications in computing.
